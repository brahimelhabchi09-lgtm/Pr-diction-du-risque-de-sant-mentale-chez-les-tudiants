```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
numeric_columns = [
    'Age', 'Year_of_Study', 'Sleep_Hours', 'Study_Hours_Per_Week', 'GPA',
    'Social_Media_Hours', 'Screen_Time_Hours', 'Financial_Stress',
    'Family_Support', 'Attendance_Rate',
]
categorical_columns = [
    'Gender', 'Physical_Activity', 'Part_Time_Job', 'Diet_Quality',
    'Previous_Mental_Health_History',
]


numeric_pip = Pipeline([
    ("Impute", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler()) 
])
categorical_pip = Pipeline([
    ('Impute' , SimpleImputer(strategy='most_freq')), 
    ('Encoder' , OneHotEncoder())
])

preprocess = ColumnTransformer([
    ('num' , numeric_pip , numeric_columns) , 
    ('cat' , categorical_pip , categorical_columns)
])

training = Pipeline([
    ('preprocess' , preprocess), 
    ('module' , LogisticRegression())
])
```
